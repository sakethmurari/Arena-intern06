```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.io as pio
pio.renderers.default = "browser"


# Load your uploaded dataset
df = pd.read_csv("sales_data (4).csv")

df.head()
```

```python
sns.set(style="whitegrid", palette="Set2")
plt.rcParams['figure.figsize'] = (8,5)
```

```python
sns.boxplot(x='Product', y='Total_Sales', data=df)
plt.title("Sales Distribution by Category")
plt.show()
```

```python
sns.violinplot(x='Product', y='Total_Sales', data=df)
plt.title("product Distribution by Total_Sales")
plt.show()
```

```python
corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
fig, axes = plt.subplots(2,2, figsize=(12,8))

# 1️⃣ Sales Distribution
sns.boxplot(x='Product', y='Total_sales', data=df, ax=axes[0,0])
axes[0,0].set_title("Sales Distribution")

# 2️⃣ Profit Spread
sns.violinplot(x='Product', y='Profit', data=df, ax=axes[0,1])
axes[0,1].set_title("Profit Spread")

# 3️⃣ Sales by Region (correct barplot)
sns.barplot(x='Region', y='Total_sales', data=df, ax=axes[1,0])
axes[1,0].set_title("Sales by Region")

# 4️⃣ Sales Histogram
sns.histplot(df['Total_sales'], kde=True, ax=axes[1,1])
axes[1,1].set_title("Sales Histogram")

plt.tight_layout()
plt.show()

```

```python
fig, axes = plt.subplots(2,2, figsize=(12,8))

sns.boxplot(x='Product', y='Total_Sales', data=df, ax=axes[0,0])
sns.violinplot(x='Product', y='Total_Sales', data=df, ax=axes[0,1])
sns.barplot(x='Region', y='Total_Sales', data=df, ax=axes[1,0])
sns.histplot(df['Total_Sales'], kde=True, ax=axes[1,1])

plt.tight_layout()
plt.show()
```

```python
df['Date'] = pd.to_datetime(df['Date'])

fig = px.line(df, x='Date', y='Total_Sales', color='Product',
              title="Interactive Sales Trend")
fig.show()
```

```python
fig = px.scatter(df, x='Price', y='Total_Sales',
                 color='Product',
                 title="Profit vs Sales Segmentation")
fig.show()
```
